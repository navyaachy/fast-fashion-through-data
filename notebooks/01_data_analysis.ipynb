### 2.2.2 Material Frequency Analysis

```python
materials = df["composition"].fillna("").str.findall(
    r"([A-Za-z]+(?:\s+[A-Za-z]+)?)\s+\d+(?:\.\d+)?%"
)

material_counts = pd.Series(
    [material for row in materials for material in row]
).value_counts()

material_counts.head(15)
```

### 2.2.3 Most Common Materials

```python
material_counts.head(10).plot(kind="bar")

plt.xlabel("Material")
plt.ylabel("Number of Mentions")
plt.title("Most Common Materials in H&M Products")
plt.xticks(rotation=45)
plt.show()
```

## 2.3 Recycled Material Analysis

```python
df["has_recycled"] = df["materials"].str.contains(
    "Recycled", case=False, na=False
)

df["has_recycled"].value_counts()
```

```python
df["has_recycled"].value_counts(normalize=True) * 100
```

### 2.3.1 Recycled Materials and Price

```python
df.groupby("has_recycled")["price"].mean()
```

```python
price_comparison = df.groupby("has_recycled")["price"].mean()

price_comparison.plot(kind="bar")

plt.xlabel("Contains Recycled Material")
plt.ylabel("Average Price")
plt.title("Average Price by Recycled-Material Mention")
plt.xticks([0, 1], ["No", "Yes"], rotation=0)
plt.show()
```

## 2.4 Product Category Analysis

```python
category_counts = df["mainCatCode"].value_counts().head(10)

category_counts
```

```python
plt.figure(figsize=(10, 5))

category_counts.plot(kind="bar")

plt.xlabel("Product Category")
plt.ylabel("Number of Products")
plt.title("Top 10 Product Categories")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
category_price = df.groupby("mainCatCode")["price"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))
category_price.plot(kind="bar")

plt.xlabel("Product Category")
plt.ylabel("Average Price")
plt.title("Average Price by Top Product Categories")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### 2.4.1 Average Price by Category Group

```python
df["mainCatCode"].str.extract(r"(ladies|men|sportswear)", expand=False).value_counts()
```

```python
gender_category = df.copy()

gender_category["gender_group"] = (
    gender_category["mainCatCode"]
    .str.extract(r"(ladies|men|sportswear)", expand=False)
)

gender_price = gender_category.groupby("gender_group")["price"].mean()

gender_price
```

```python
plt.figure(figsize=(7, 5))

gender_price.plot(kind="bar")

plt.xlabel("Category Group")
plt.ylabel("Average Price")
plt.title("Average Price by Category Group")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 3. Initial Findings

- The dataset contains a wide range of H&M products, with prices concentrated toward the lower end.
- Polyester, cotton and spandex are among the most frequently mentioned materials in product compositions.
- Around 60.65% of the product records contain a mention of recycled material.
- Products with a recycled-material mention have a higher average listed price than products without one.
- Product categories differ considerably in both product count and average listed price.
- Among the identified category groups, sportswear has the highest average listed price, followed by men’s and ladies’ products.
